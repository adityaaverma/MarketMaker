```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('../prices.csv')
```

```python
plt.plot(df['step'], df['price'])
plt.xlabel('Step')
plt.ylabel('Price')
plt.title('GBM simulated Price path')
plt.show()
```

```python
df2=pd.read_csv('../pricesBA.csv')
```

```python
plt.figure(figsize=(10,5))
plt.plot(df2['step'],df2['price'],label='bid',color='black')
plt.plot(df2['step'],df2['bid'],label='bid',linestyle='--')
plt.plot(df2['step'],df2['ask'],label='ask',linestyle='--')
plt.xlabel('Step'); plt.ylabel('Price')
plt.title('Naive Quoter: bid/ask around price')
plt.legend()
plt.show()
```

```python
spread=df2['ask']-df2['bid']
print(spread.describe())

print((df2['bid']<df2['price']).all())
print((df2['ask']>df2['price']).all())
```
